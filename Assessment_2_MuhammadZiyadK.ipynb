```python
#Q1. NumPy Array Creation – 2 Marks

array_1 = np.array([10,20,30,40,50])
print(array_1)
print("Shape:",array_1.shape)
print("Data type:",array_1.dtype)
```

```python
#Q2. NumPy Array Operations – 2 Marks
arr = np.array([5, 10, 15, 20, 25])

#Add 5 to every element in the array
new_arr = arr + 5
print ("1: ",new_arr)

#Multiply every  element by 2.
arrx2 = arr * 2
print("2: ",arrx2)

```

```python
#Q3. NumPy Statistical Operations
arr = np.array([12, 15, 18, 20, 25, 30])

#Mean
mean = np.mean(arr)
print("Mean:",mean)

#Median
median = np.median(arr)
print("Median:",median)

#Minimum value
min = np.min(arr)
print("Minimum value:",min)

#Maximum Value
max = np.max(arr)
print("Maximum value:",max)

#Standard devaiation
std = np.std(arr)
print("Standard deviation:",std)


```

```python
#Q4. NumPy Filtering – 2 Marks
arr = np.array([10, 25, 30, 45, 50, 65, 70])

#Using NumPy, extract all values that are greater than 40.
print("greater_than_40",arr[arr > 40])

```

```python
#Q10. Handling Missing Values
df = pd.DataFrame({
'Name': ['A', 'B', 'C', 'D'],
'Age': [21, np.nan, 23, 22],
'Score': [80, 75, np.nan, 90]
})
#Identify the missing values
missing_values = df.isnull()
print(missing_values)
#Replace the missing Age with the mean age
age_mean = df['Age'].mean()
df['Age'].fillna(age_mean, inplace=True)
print(df)
```
